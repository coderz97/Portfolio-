# 1. Linear Regression

Linear regression is a statiscical concept that helps find the best line that represents the relationship between features and target and also minimise difference between actual and predicted values.

y=mx+b

Types of Linear Regression
1. Simple Linear Regression
2. Multiple Linear Regression

Linear Regression methods/models (to estimate coefficients)
1. OLS - Ordinary Least Squares (standard method, no penalty)
2. Lasso Regression /L1 Regularization (Least Absolute Shrinkage and Selection Operator)
3. Ridge Regression /L2 Regularization 
4. Gradient Descent 

Lasso, Ridge and Gradient Descent are regularization/optimization methods.


### Ordinary Least Squares(OLS)
Aims to locate the line (or hyperplane,in case of multiple variables) that minimizes the sum of squared errors between actual and predicted values of target variable.
 
        equation: min RSS = ∑(y-y')^2

### Lasso / L1 Regularization (Least Absolute Shrinkage and Selection Operator)
builds on OLS by including a penalty term 
removes less important features by setting their coeffients to 0 (can be used as feature selection)
prevents overfitting

        equation: min RSS+λ×∑β 

### Ridge Regression/ L2 Regularization 
builds on OLS by including a different penalty term, alternative to lasso
shrinks the coefficients of less important features.  (does not set to 0)
prevents overfitting

        equation: min RSS+λ×∑β^2

### Gradient Descent
iteratively finds the weights and bias that produce the model with the lowest loss.
