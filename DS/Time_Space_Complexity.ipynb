## Time Complexity

Time complexity is the computational complexity that describes the number of operations an algorithm performs relative to the input size.

In short

Time complexity measures how the number of basic operations grows with the size of the input

DS itself does not have a time complexity, instead the time complexity depends on the operation performed on the DS.

Big O represents worst case scenerio

0(5n^3+n^2+52)--> O(n^3) just take the biggest one

### O(1)---> if number of operations does not depend on the input size
eg:
 Print("Hello")
 access an array element using index , a[1]


### O(n)---> number of operations increases linearly with input size
eg:
linear search on an array with n elements
for i in range(n); print a[n]

### O(n^2)---> number of operations increase n^2 times based on input size
eg:
nested for loops , for (n); for (n)
traversing / linear search on n*n sized matrix

### O(log n)--->  number of operations increse as a log fn of input size

better than O(n), bcz as input size grows, logarithmic time increases much more slowly than linear time.

eg: Binary search on array of size n

when finding all permutations
